# Python Directory Management


### os and os.path module

```python
import os
os.mkdir("./RequireFiles/my_directory")
os.makedirs("./RequireFiles/parent_directory/child_directoy")
```

##### Getting Current Working Directory (CWD)

```python
import os
print("String format: ", os.getcwd())
print("Byte string format: ", os.getcwdb())
```

##### Renaming a directory

```python
os.rename("./RequireFiles/my_directory", "renamed_directory")

os.renames("./RequireFiles/parent_directory", "./RequireFiles/new_parent/new_child/renamed_directory")
```

##### Changing Current Working Directory (CWD)

```python
print("Before changing directory: ", os.getcwd())
os.chdir("./RequireFiles")


print("after changing directory: ", os.getcwd())
```

##### Listing Files in a Directory

```python
import os
print("Files in CWD are: ", os.listdir(os.getcwd()))
```

##### Removing a directory

```python
import os
dir_path = "D:/New folder/Courses/PythonSeries/02-Intermidate_Python/RequireFiles"
if not os.listdir(dir_path):
    os.rmdir(dir_path)
    print("Cirectory removed successfully")
else:
    print("Error! Directory not empty.")
```

##### Check Whether It Is a Directory

```python
import os
cwd = "/"
print(os.path.isdir(cwd))

other = "D:/"
print(os.path.isdir(other))
```
