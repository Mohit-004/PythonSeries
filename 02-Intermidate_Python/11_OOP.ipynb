### Python Objects

```python
class Dog:
    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.species)
```

### __ init __ Method

__init__ method is the constructor in Python, automatically called when a new object is created. It initializes the attributes of the class.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)
print(dog1.name)
```

### Class and Instance Variables

```python
class Dog:
    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)
dog2 = Dog("Charlie", 3)


print(dog1.species)
print(dog1.name)

dog1.name = "Max"
print(dog1.name)


Dog.species = "Feline"

print(dog1.species)
print(dog2.species)
```
