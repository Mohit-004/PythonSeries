### Assigning Values to Variables

```python
#1. Basic Assignment: Variables are assigned values using the = operator.
x = 5
y = 3.14
z = "Hii"
```

```python
#2. Dynamic Typing: Python is dynamically typed, so the same variable can store different data types during execution.
x = 10
x = "Now a string"
```

```python
#3. Assigning Same Value: same value can be assigned to multiple variables in a single line.
a = b = c = 100
print(a, b, c)
```

```python
#4. Assigning Different Values: Multiple variables can also be assigned different values in a single line.
x, y, z = 1, 2.5, "Python"
print(x, y, z)
```

### Concept of Object Reference

```python
x = 5
y = 5
print(x, y)

x = "geeks"
y = "Computer"
#Here x and y refers to other values but 5 remains in memory it is not deleted
print(x, y)
```

### Understanding Variable Reassignment

```python
x = 1
y = x
y = y + 1

print(x)
print(y)
```

### Deleting a Variable

```python
x = 10
del x
print(x)
```

### Practical Examples

```python
#1. Swapping Two Variables
a, b = 3, 5
a, b = b, a
print(a, b)
```

```python
# 2. Counting Characters in a String
word = "Python"
length = len(word)
print(word)
print(length)
```
