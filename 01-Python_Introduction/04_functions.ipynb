## Function Arguments

```python
def evenOdd(x):
    if (x % 2 == 0):
        return "Even"
    else:
        return "Odd"
    
print(evenOdd(16))
print(evenOdd(7))
```

### Types of Function Arguments

```python
# 1. Default argument: Default argument use a predefined value when no value is passed during the function call.
def myFun(x, y = 50):
    print("X: ", x)
    print("Y: ", y)

myFun(10)
```

```python
# 2. Keyword Arguments: pass values using parameter names, so argument order does not matter.
def student(fname, lname):
    print(fname, lname)

student(fname="Geeks", lname = "Practice")
student(lname="Practice", fname="Geeks")
```

```python
# 3. Positional Arguments: values are assigned to parameters based on their order in the function call.
def nameAge(name, age):
    print("Hi, I am", name)
    print("My age is ", age)

print("Case-1:")
nameAge("Olivia: ", 27)

print("Case-2:")
nameAge(27, "Olivia")
```

```python
# 4. Arbitrary Arguments: allow functions to accept multiple values. This is done using two special symbols:

# *args collects extra positional arguments as a tuple.
# **kwargs collects extra keyword arguments as a dictionary.

def myFun(*args, **kwargs):
    print("Non-Keyword Arguments {*args}: ")
    for arg in args:
        print(arg)

    print("Keyword Arguments (**kwargs): ")
    for key, value in kwargs.items():
        print(f"{key} == {value}")

myFun("Hey", "Welcome", first="Geeks", mid="for", last="Geeks")
```

```python
# Function within Functions
def f1():
    s = "I love geeksforgeeks"
    def f2():
        print(s)

    f2()
f1()
```

# Global and Local Variables in Python

### Local Variables

```python
def greet():
    msg = "Hello from inside the function!"
    print(msg)

greet()
```

```python
def greet():
    msg = "Hello!"
    print("Inside function: ", msg)

greet()
print("Outside function: 0", msg)
```

### Global Variables

```python
msg = "Python is awesome!"

def display():
    print("Inside function: ", msg)
```

### Use of Local and Global variables

```python
def fun():
    s = "Me too."
    print(s)

s = "I love GeeksForGeeks"

fun()
print(s)
```

### Modifying Global Variables Inside a Function

```python
def fun():
    s += "GFG"
    print(s)

s = "I love GeeksforGeeks"
fun()
```

```python
s = "Python is great!"

def fun():
    global s
    s += "GFG"
    print(s)
    s = "Look for GeeksForGeeks python Section"
    print(s)

fun()
print(s)
```

```python
a = 1

def f():
    print("f():", a)

def g():
    a = 2
    print("g():", a)

def h():
    global a
    a = 3

    print("h(): ", a)

print("global:", a)
f()
print("global:", a)
g()
print("global:", a)
h()
print("global:", a)
```
