# Control Flow Statements

## 1. Loops in Python

### For Loop

```python
n = 4 
for i in range(0, n):
    print(i)
```

```python
# Iterating by Index of Sequences
a = ["geeks", "for", "geeks"]
for idx in range(len(a)):
    print(a[idx])
```

### While Loop

```python
cnt = 0
while (cnt < 3):
    cnt = cnt + 1
    print("Hello Geek")
```

```python
# Infinite While Loop
while True:
    print("Hello Geek")
```

### Nested Loops

```python
for i in range(1, 5):
    for j in range(i):
        print(i, end=" ")
    print()
```
