# Control Flow Statements

## 1. Loops in Python

### For Loop

```python
n = 4 
for i in range(0, n):
    print(i)
```

```python
# Iterating by Index of Sequences
a = ["geeks", "for", "geeks"]
for idx in range(len(a)):
    print(a[idx])
```

### While Loop

```python
cnt = 0
while (cnt < 3):
    cnt = cnt + 1
    print("Hello Geek")
```

```python
# Infinite While Loop
while True:
    print("Hello Geek")
```

### Nested Loops

```python
for i in range(1, 5):
    for j in range(i):
        print(i, end=" ")
    print()
```

## Using Else with Loops in Python

#### The else block just after for/while is executed only when the loop is NOT terminated by a break statement.

```python

for i in range(1, 4):
    print(i)
else:
    print("No Break")
```

```python
for i in range(1, 4):
    print(i)
    break
else:
    print("No Break")
```

```python
# Python 3.x program to check if an array consists 
# of even number
number = [1,2,3,9,5]
for i in range(len(number)):
    if number[i] % 2 == 0:
        print("List contains even numbers.")
        break
else:
    print("List does not contains even numbers.")
```

```python
count = 0
while count < 1:
    count = count + 1
    print(count)
    break
else:
    print("No break")
```

## Practice Problems of Loops in Python

```python
# Problem 1: Print each item in a shopping list
items = input("Enter shopping items seperated by commas: ").split(',')
for item in items:
    print("Buy: ", item.strip())
```

```python
#Problem 2: Print squares of numbers from 1 to n
n = int(input("Enter a number: "))
for i in range(1,n+1):
    print(f"Square of {i} is : ", i**2)
```

```python
# Problem 3: Countdown timer
seconds = int(input("Enter the countdown time in seconds: "))
while seconds > 0:
    print(f"Time Left : ", seconds)
    seconds -= 1
```

```python
# Problem 4: Sum until user enters 0
total = 0
number = int(input("Enter the number (o to stop): "))
while number != 0:
    total += number
    print("Sum is : ", total)
    number = int(input("Enter the number (o to stop): "))
```

```python
# Problem 5: Print a multiplication table
n = int(input("Enter the number want to calculate multiplication table till"))

for num in range(1,n+1):
    for i in range(1,11):
        print(num*i, end=" ")

    print()
```

```python
# Problem 6: Print Identity Matrix Pattern
n = 4
for i in range(n):
    for j in range(n):
        if i == j:
            print("1", end=" ")
            continue
        print("0", end=" ")
    print()
```

```python
# Problem 7: Stop printing at a target item
items = ["apple", "banana", "cherry", "stop", "mango"]

for item in items:
    if item == "stop":
        break
    print("Item: ", item)
```

```python
# Problem 2: Find first even number
while True:
    num = int(input("Enter the number: "))
    if num % 2 == 0:
        print("First even number is found: ", num)
        break
```

```python
# Problem 1: Skip out-of-stock items
items = ["milk", "bread", "out of stock", "eggs"]

for item in items:
    if item == "out of stock":
        continue
    print("Item: ", item)
```

```python
# Problem 2: Skip even numbers
n = 10

for i in range(1, n+1):
    if i % 2 == 0:
        continue
    print("Odd Number: ", i)
```

```python
# Problem 1: Use pass for future implementation
tasks = ["email", "meeting", "call"]

for task in tasks:
    if task == "call":
        pass
    print("Do: ", task)
```
