### Truthy vs Falsy Values in Python

```python
number = 7

if number:
    print("This will print because 7 is truthy.")

number = 0

if number:
    print("This will not print because 0 is falsy.")
```

```python
#Truthy Values
# These evaluate to True in a Boolean context:

# Non-empty sequences or collections: [ 1 ], ( 0, ), "Hello", { 1:2 }
# Numeric values not equal to zero: 1, -4, 3.5
# Constant: True
if [1, 2]: 
    print("Non empty list is truthy.")

if -4:
    print("-4 is truthy.")
```

```python
# Falsy Values
# These evaluate to False in a Boolean context:

# Empty sequences and collections: [ ], ( ), { }, set( ), " ", range(0)
# Numbers: 0 (integer), 0.0 (float), 0j (complex)
# Constants: None, False

if not 0:
    print("0 is falsy.")

if not []:
    print("Empty list is falsy.")
```

```python
# Example: This program demonstrates how truthy and falsy values can simplify conditions directly in the main code.

num1 = 7
num2 = 4

if num1 %2 :
    print(num1, " is odd")
else:
    print(num1, " is even")


if num2 % 2:
    print(num2, " is odd")
else:
    print(num2, " is even")
```

## 4) Bitwise Operators

```python
# Bitwise operators act on bits and perform bit-by-bit operations. These are used to operate on binary numbers.
# Bitwise Operators in Python are as follows:

a = 10
b = 4

print(a & b) # Bitwise AND
print(a | b) # Bitwise OR
print(~a) # Bitwise NOT
print(a ^ b) # Bitwise XOR
print(a >> 2) # Bitwise Shift
print(a << 2)
```

## Python := Walrus Operator in Python 3.8

```python
#The Walrus Operator (:=), introduced in Python 3.8, allows you to assign a value to a variable as part of an expression. It helps avoid redundant code when a value needs to be both used and tested in the same expression — especially in loops or conditional statements.
num = [1,2,3,4,5]

while (n:= len(num)) > 0:
    print(num.pop())
```

```python
d = [
    {"userId": 1, "name": "rahul", "completed": False},
    {"userId": 1, "name": "rohit", "completed": False},
    {"userId": 1, "name": "ram", "completed": False},
    {"userId": 1, "name": "ravan", "completed": True}
]

print("With Python 3.8 Walrus operator: ")
for entry in d:
    if name := entry.get("name"):
        print(name)

print("Without Walrus operator: ")
for entry in d:
    name = entry.get("name")
    if name: 
        print(name)
```

```python
#Example 3: Simplifying User Input Loops
foods = []
while True:
    f = input("What food do you like?: ")
    if f == "quit":
        break;
    foods.append(f)
```

```python
#With Walrus Operator
foods = []

while (f := input("What food do you like? (type 'quit' to stop): ")) != "quit":
    foods.append(f)
```

# 2. Conditional Statements in Python

## If Statement

```python
age = 20
if age >= 18:
    print("Eligible to vote.")
```

```python
age = 19
if age > 18: print("Eligible to vote.")
```

## If else Statement

```python
age = 10
if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
```

## If-elif-else Statement

```python
age = 25

if age <= 12:
    print("Child.")
elif age <=19:
    print("Teenager.")
elif age <= 35:
    print("Young adult.")
else:
    print("Adult.")
```

## Nested if-else Statement

```python
age = 70
is_member = True

if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senoir discount.")
```

## Ternary Conditional Statement

```python
age = 20
s = "Adult" if age >= 18 else "Minor"
print(s)
```

## Match-Case Statement

```python
number = 8

match number:
    case 1:
        print("One")
    case 2|3:
        print("Two or Three")
    case _:
        print("Other number")
```

```python
# Problem 13: Activity Suggestion based on weather condition
weather = input("enter the weather (sunny/rainy/cloudy/snowy): ").lower()
match weather:
    case "sunny":
        print("Great day for a picnic!")
    case "rainy":
        print("Stay indoors and read a book.")
    case "cloudy":
        print("Perfect time for a walk.")
    case "snowy":
        print("Build a snowman or go skiing!")
    case _:
        print("Unknown weatherr condition.")
```
