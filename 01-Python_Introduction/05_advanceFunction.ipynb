# Recursion in Python

```python
def factorial(n):
    if n == 0:
        return 1
    else :
        return n * factorial(n-1)

print(factorial(5))
```

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))
```

# Types of Recursion

### Tail Recursion: The recursive call is the last thing the function does, so nothing happens after it returns. Some languages can optimize this to work like a loop, saving memory.

### Non-Tail Recursion: The function does more work after the recursive call returns, so it can’t be optimized into a loop.

```python
def tail_fact(n , acc=1):
    if n == 0:
        return acc
    else:
        return tail_fact(n-1, acc * n)

def nontail_fact(n):
    if n == 0:
        return 1
    else:
        return n * nontail_fact(n-1)

print(tail_fact(5))
print(nontail_fact(5))
```

# *args and **kwargs in Python

```python
def fun( *args):
    return sum(args)

print(fun(5, 10, 15))

def fun(**kwargs):
    for k, val in kwargs.items():
        print(k, val)

fun(a = 1, b = 2, c = 3)
```

# Symbols for Handling Variable Arguments

### 1. Non-Keyword Arguments (*args)

```python
def myFun(*argv):
    for arg in argv:
        print(arg)

myFun("Hello", "Welcome", "to", "GeeksforGeeks")
```

```python
def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result

print(multiply(2, 3, 4))
```

### 2. Keyword Arguments (**kwargs)

```python
def fun(**kwargs):
    for k , val in kwargs.items():
        print(k, "=", val)

fun(s1="Python", s2="is", s3="Awesome")
```

```python
def introduce(**kwargs):
    details = []
    for k,v in kwargs.items():
        details.append(k + ":" + str(v))
    return ",".join(details)

print(introduce(Name="Alice", Age= 25, City="New York"))
```

### Using both *args and **kwargs

```python
def student_info(*args, **kwargs):
    print('Subjects:', args)
    print("Details:", kwargs)

student_info("Math", "Science", "English", Name="Alice", Age = 20, City="New York")
```

# Python Lambda Functions

```python
a = "GeekForGeeks"
upper = lambda x: x.upper()
print(upper(a))
```

### Use Cases

```python
# 1. Condition Checking: lambda function can use conditional expressions (if-else) to return different results based on a condition.
check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(check(5))
print(check(-3))
print(check(0))
```

```python
# 2. List Comprehension: Lambda can be combined with list comprehensions to apply the same operation to multiple values in a compact way.
func = [lambda arg=x: arg * 10 for x in range(1, 5)]
for i in func:
    print(i())
```

```python
# 3. Returning Multiple Results: Although a lambda can contain only one expression, it can still return multiple results by combining them into a tuple.

calc = lambda x, y: (x + y, x * y)
res = calc(3, 4)
print(res)



```

```python
# 4. filter(): This function uses a lambda expression to select elements from a list that satisfy a given condition, such as keeping only even numbers.
c = [1, 2, 3, 4, 5, 6]
even = filter(lambda x: x % 2 == 0, c)
print(list(even))
```

```python
# 5. map(): This function applies a lambda expression to each element and returns a map object. It can be converted to a list using list().
a = [1, 2, 3, 4]
double = map(lambda x: x * 2, a)
print(list(double))
```

```python
# 6. reduce(): This function repeatedly applies a lambda expression to elements of a list to combine them into a single result.
from functools import reduce
a = [1,2,3,4]
mul = reduce(lambda x, y: x * y, a)
print(mul)
```

# Python map() function

```python
s = ['1', '2', '3', '4']
res = map(int, s)
print(list(res))
```

### Converting map object to a list


```python
def double(val):
    return val * 2

a = [1, 2, 3, 4]
res = list(map(double, a))
print(res)
```

### map() with lambda

```python
a = [1, 2, 3, 4]
res = list(map(lambda x:x ** 2, a))
print(res)

```

### map() with multiple iterables

```python
a = [1, 2, 3]
b = [4, 5, 6]
res = map(lambda x, y: x + y, a, b)
print(list(res))
```

### Converting strings to Uppercase

```python
fruits = ["apple", "banana", "cherry"]
res = map(str.upper, fruits)
print(list(res))
```

### Extracting first character from strings

```python
words = ["apple", "banana", "cherry"]
res = map(lambda s : s[0], words)
print(list(res))
```

```python
s = ['  hello   ', '    world   ', '    python  ']
res = map(str.strip, s)
print(list(res))
```

### Calculate fahrenheit from celsius

```python
celsius = [0, 20, 37, 100]
fahrenheit = map(lambda c: (c * 9/5) + 32, celsius)
print(list(fahrenheit))
```
